```python
import os
import re
modsurls = {}
archive_urls = {}
hathi_urls = {}
slub_urls = {}
gbooks_urls = {}

def additem(curitem,curdict):
    if(curitem in curdict):
        curdict[curitem] = curdict[curitem] + 1
    else:
        curdict[curitem] = 1

def getpdfs(fname):

    global modsurls
    f=open(fname)
    for l in f:
        m = re.search('<mods:url[^>]*>([^<]+)',l)
        if(m):
            cururl = m[1]
            if(re.search('archive.org',cururl)):
                additem(cururl,archive_urls)
            if(re.search('books.google',cururl)):
                additem(cururl,gbooks_urls)
            if(re.search('hdl.handle.net',cururl)):
                additem(cururl,hathi_urls)
            if(re.search('slub-dresden',cururl)):
                additem(cururl,slub_urls)
            if(cururl in modsurls):
                modsurls[cururl] = modsurls[cururl] + 1
            else:
                modsurls[cururl] = 1
    f.close()
```

```python
greektextgroups = {}
works = {}
greekworks = {}
latinworks = {}
englishworks = {}
latxls = {}
grcxls = {}
totmods = 0
for top, dirs, files in os.walk(os.getcwd()):
    #print("Printing directories...")
    #for dir in dirs:
    #    print(os.path.join(top, dir))
    #print("Printing files....")
    #print(len(files))
    for file in files:
        if(not re.search('\.xml$',file)):
            continue
            
        getpdfs(os.path.join(top, file))
        totmods = totmods + 1
        args = file.split('.')
        curtextgroup = args[0]
        curwork = args[0] + '.' + args[1]
        curedition = curwork + '.' + args[2]

        if(re.search('eng',args[2]) ):
            if(re.search('latinLit',os.path.join(top, file))):
                latxls[file] = 1
            if(re.search('greekLit',os.path.join(top, file))):
                grcxls[file] = 1
                
            englishworks[file] = 1
        elif(re.search('grc',args[2])):
            if(curtextgroup in greektextgroups):
                greektextgroups[curtextgroup] = greektextgroups[curtextgroup] + 1
            else:
                greektextgroups[curtextgroup] =  1
            if(curwork in greekworks):
                greekworks[curwork] = greekworks[curwork] + 1
            else:
                greekworks[curwork] =  1
            greekworks[file] = 1
        elif(re.search('lat',args[2])):
            latinworks[file] = 1
        #print(file,os.path.join(top, file))
othereng = len(englishworks)-(len(grcxls)+len(latxls))
print('totmods',totmods,'toturls',len(modsurls),'archive.org',len(archive_urls),'google',len(gbooks_urls),'hathi',len(hathi_urls),'slub',len(slub_urls))
print('all PDFs',len(archive_urls)+len(gbooks_urls)+len(hathi_urls)+len(slub_urls))
print('greek xlations',len(grcxls),'latin xlations',len(latxls),'other',othereng)
print('greektextgroups',len(greektextgroups),'greek',len(greekworks),'latin',len(latinworks),'english',len(englishworks))
i = 0
for foo in sorted(greekworks,key=greekworks.get,reverse=True):
    i = i + 1
    if(i>10):
        break
    print(greekworks[foo],foo)
i = 0
for foo in sorted(greektextgroups,key=greektextgroups.get,reverse=True):
    i = i + 1
    if(i>10):
        break
    print(greektextgroups[foo],foo)
```
